```python
from Bio import SeqIO
import numpy as np
import sys
sys.path.append('..')
from src.discrimalign import discrimalign
rng = np.random.default_rng()
```

```python
BLOCKS_raw = list(SeqIO.parse('blocks_5.0.fa', 'fasta'))
```

```python
BLOCKS = {}
for s in BLOCKS_raw:
    header = s.id.split('|')
    block_id = header[0]
    prot_id = header[1]
    s.id = block_id + '|' + prot_id
    s.name = ''
    s.description = ''
    try:
        BLOCKS[block_id].append(s)
    except KeyError:
        BLOCKS[block_id] = [s]
```

```python
block_names = list(BLOCKS)
```

```python
nbs_of_pairs = [len(BLOCKS[b])*(len(BLOCKS[b])-1)/2 for b in BLOCKS]
```

```python
sum(nbs_of_pairs)
```

```python
positive_pairs = []
for b in BLOCKS:
    seqlist = BLOCKS[b]
    for i, s in enumerate(seqlist):
        for s2 in seqlist[(i+1):]:
            positive_pairs.append((s, s2))
len(positive_pairs)
```

```python
negative_pairs = []
for _ in range(len(positive_pairs)):
    b1, b2 = rng.choice(block_names, 2, replace=False)
    s1 = BLOCKS[b1][rng.choice(len(BLOCKS[b1]))]
    s2 = BLOCKS[b2][rng.choice(len(BLOCKS[b2]))]
    negative_pairs.append((s1, s2))
```

```python
seqlistA = [l[0] for l in positive_pairs] + [l[0] for l in negative_pairs]
seqlistB = [l[1] for l in positive_pairs] + [l[1] for l in negative_pairs]
labels = [1]*len(positive_pairs) + [0]*len(negative_pairs)
```

```python
discrblosum = discrimalign(seqlistA, seqlistB, labels, 
                          aligner_mode = 'local',
                          gap_mode='affine',
                          substitution_mode = 'symmetric',
                          num_threads=12,
                           verbose=True
                          )
```
